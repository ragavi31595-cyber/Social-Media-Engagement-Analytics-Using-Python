```python
import pandas as pd
import numpy as np
import re

# --------------------------------
# 1. Missing Value Detection
# --------------------------------

print("Missing Values Before Cleaning:")
print(df.isnull().sum())


# --------------------------------
# 2. Handle Missing Values
# --------------------------------

# Numeric columns - fill with median
numeric_columns = df.select_dtypes(include='number').columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

# Categorical columns - fill with mode
categorical_columns = df.select_dtypes(include='object').columns

for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])


# --------------------------------
# 3. Remove Duplicates
# --------------------------------

print("Duplicate rows before removal:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate rows after removal:", df.duplicated().sum())


# --------------------------------
# 4. Data Formatting
# --------------------------------

# Convert posted_at to datetime
df['posted_at'] = pd.to_datetime(df['posted_at'], errors='coerce')

# Convert numeric columns
numeric_cols = [
    'age',
    'likes',
    'comments',
    'shares',
    'watch_time_sec',
    'impression_count',
    'follower_count',
    'engagement_rate'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')


# --------------------------------
# 5. Standardize Gender
# --------------------------------

df['gender'] = df['gender'].astype(str).str.strip().str.lower()

df['gender'] = df['gender'].replace({
    'm': 'male',
    'f': 'female',
    'o': 'other'
})


# --------------------------------
# 6. Correct Unrealistic Values
# --------------------------------

engagement_cols = ['likes', 'comments', 'shares']

for col in engagement_cols:
    df.loc[df[col] < 0, col] = np.nan
    df[col] = df[col].fillna(df[col].median())


# --------------------------------
# 7. Extract Hashtag Count
# --------------------------------

def count_hashtags(text):
    if pd.isna(text):
        return 0
    return len(re.findall(r'#\w+', str(text)))

df['hashtag_count'] = df['hashtags'].apply(count_hashtags)


# --------------------------------
# 8. Clean Sentiment Labels
# --------------------------------

df['sentiment'] = df['sentiment'].astype(str).str.strip().str.lower()

df['sentiment'] = df['sentiment'].replace({
    'pos': 'positive',
    'neg': 'negative',
    'neu': 'neutral'
})


# --------------------------------
# Final Check
# --------------------------------

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nFinal Dataset Shape:", df.shape)

print("\nGender Values:")
print(df['gender'].value_counts())

print("\nSentiment Values:")
print(df['sentiment'].value_counts())

print("\nFinal Dataset:")
display(df.head())
```

```python
# TASK 4 — DATA WRANGLING
# ==========================================

import numpy as np
import re

# ------------------------------------------
# 1. Create Engagement Score
# ------------------------------------------

df['engagement_score'] = (
    df['likes'] +
    df['comments'] +
    df['shares']
)

# ------------------------------------------
# 2. Create Hashtag Count
# ------------------------------------------

def count_hashtags(text):
    if pd.isna(text):
        return 0
    return len(re.findall(r'#\w+', str(text)))

df['hashtag_count'] = df['hashtags'].apply(count_hashtags)


# ------------------------------------------
# 3. Log-Transformed Metrics
# ------------------------------------------

df['log_likes'] = np.log1p(df['likes'])
df['log_comments'] = np.log1p(df['comments'])
df['log_shares'] = np.log1p(df['shares'])
df['log_impressions'] = np.log1p(df['impression_count'])


# ------------------------------------------
# 4. GroupBy — Post Type
# ------------------------------------------

post_type_summary = df.groupby('post_type').agg({
    'likes': 'mean',
    'comments': 'mean',
    'shares': 'mean',
    'impression_count': 'mean',
    'engagement_score': 'mean',
    'engagement_rate': 'mean'
}).round(2)

print("Post Type Summary:")
display(post_type_summary)


# ------------------------------------------
# 5. GroupBy — Country
# ------------------------------------------

country_summary = df.groupby('country').agg({
    'likes': 'mean',
    'comments': 'mean',
    'shares': 'mean',
    'impression_count': 'mean',
    'engagement_score': 'mean',
    'engagement_rate': 'mean'
}).round(2)

print("Country Summary:")
display(country_summary)


# ------------------------------------------
# 6. GroupBy — Sentiment
# ------------------------------------------

sentiment_summary = df.groupby('sentiment').agg({
    'likes': 'mean',
    'comments': 'mean',
    'shares': 'mean',
    'impression_count': 'mean',
    'engagement_score': 'mean',
    'engagement_rate': 'mean'
}).round(2)

print("Sentiment Summary:")
display(sentiment_summary)


# ------------------------------------------
# 7. Check New Fields
# ------------------------------------------

print("New Columns:")
print([
    'engagement_score',
    'hashtag_count',
    'log_likes',
    'log_comments',
    'log_shares',
    'log_impressions'
])

print("\nUpdated Dataset:")
display(df.head())
```

```python
# TASK 5 — STATISTICAL ANALYSIS
# ==========================================

# Select required columns
stats_cols = [
    'likes',
    'comments',
    'shares',
    'watch_time_sec',
    'engagement_rate',
    'follower_count'
]

stats_df = df[stats_cols]


# ------------------------------------------
# 1. Mean
# ------------------------------------------

mean_values = stats_df.mean()

print("MEAN")
display(mean_values.to_frame('Mean'))


# ------------------------------------------
# 2. Median
# ------------------------------------------

median_values = stats_df.median()

print("MEDIAN")
display(median_values.to_frame('Median'))


# ------------------------------------------
# 3. Mode
# ------------------------------------------

mode_values = stats_df.mode().iloc[0]

print("MODE")
display(mode_values.to_frame('Mode'))


# ------------------------------------------
# 4. Standard Deviation
# ------------------------------------------

std_values = stats_df.std()

print("STANDARD DEVIATION")
display(std_values.to_frame('Standard Deviation'))


# ------------------------------------------
# 5. Variance
# ------------------------------------------

variance_values = stats_df.var()

print("VARIANCE")
display(variance_values.to_frame('Variance'))


# ------------------------------------------
# 6. Percentiles
# ------------------------------------------

percentiles = stats_df.quantile([0.10, 0.25, 0.50, 0.75, 0.90])

print("PERCENTILES")
display(percentiles)


# ------------------------------------------
# 7. Skewness
# ------------------------------------------

skewness = stats_df.skew()

print("SKEWNESS")
display(skewness.to_frame('Skewness'))


# ------------------------------------------
# 8. Kurtosis
# ------------------------------------------

kurtosis = stats_df.kurtosis()

print("KURTOSIS")
display(kurtosis.to_frame('Kurtosis'))
```

```python
# 3. Behavioral Insights
# A. Best time of day for impressions

# First extract the hour:

df['hour'] = df['posted_at'].dt.hour

# Create time periods:

def get_time_period(hour):
    if 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'

df['time_of_day'] = df['hour'].apply(get_time_period)
```

```python
time_impressions = (
    df.groupby('time_of_day')['impression_count']
      .mean()
      .sort_values(ascending=False)
      .round(2)
)

print("Average Impressions by Time of Day:")
display(time_impressions)
```

```python
print("Highest Impression Time:",
      time_impressions.index[0])

print("Average Impressions:",
      time_impressions.iloc[0])
```

Insight format:

Time of Day: [Night] recorded the highest average impressions at approximately [50013.73].

```python
# B. Device type impact on watch time
device_watch = (
    df.groupby('device_type')['watch_time_sec']
      .mean()
      .sort_values(ascending=False)
      .round(2)
)

print("Average Watch Time by Device:")
display(device_watch)
```
